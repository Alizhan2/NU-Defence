# GeoWatch AI — DOTA8 smoke training
YOLOv8n-OBB: 4 train + 4 val images. Pipeline proof only, not production accuracy evidence.

```python
!pip -q install 'ultralytics>=8.3,<9'
from ultralytics import YOLO
import json, pathlib
!nvidia-smi || true
```

```python
model=YOLO('yolov8n-obb.pt')
model.train(data='dota8.yaml',epochs=20,imgsz=1024,batch=4,device=0,seed=42,deterministic=True,project='/content/geowatch_runs',name='dota8_smoke',exist_ok=True)
```

```python
best='/content/geowatch_runs/dota8_smoke/weights/best.pt'
trained=YOLO(best)
m=trained.val(data='dota8.yaml',imgsz=1024,device=0)
report={'dataset':'DOTA8 (4 train / 4 val)','purpose':'pipeline smoke test only','model':'yolov8n-obb','epochs':20,'map50':float(m.box.map50),'map50_95':float(m.box.map),'precision':float(m.box.mp),'recall':float(m.box.mr)}
pathlib.Path('/content/metrics.json').write_text(json.dumps(report,indent=2))
print(json.dumps(report,indent=2))
```
